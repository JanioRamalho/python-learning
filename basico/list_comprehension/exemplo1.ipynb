## Um exemplo prático de List Comprehension

### O que faríamos se quisermos ordenar 2 listas "relacionadas"

```python
vendas_produtos = [1500, 150, 2100, 1950]
produtos = ["Vinho" , "Cafeteira", "Microondas" , "Iphone"]


lista_aux = list(zip(vendas_produtos, produtos))
lista_aux.sort(reverse=True)

#Vamos agora criar uma lista apenas com os produtos, mas ordenados de acordo com o valor das vendas.
produtos = [produto for venda, produto in lista_aux]
print(produtos)
```
