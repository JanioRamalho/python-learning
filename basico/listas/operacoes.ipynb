# Métodos de listas

Neste arquivo vou testar alguns métodos importantes de listas:

- `insert()`
- `pop()`
- `index()`
- `count()`
- `sort()`
- `reverse()`
- `extend()`

```python
produtos_eletronicos = [
    "Computador",
    "Celular",
    "Televisão",
    "PlayStation",
    "Geladeira",
    "Nobreak"
]
```

## 1. insert()

O `insert()` adiciona um objeto em uma posição escolhida da lista.

Quando adicionamos um novo valor em um índice que já estava ocupado, o elemento antigo não é apagado. Ele passa para a posição seguinte.

A estrutura é:

`lista.insert(indice, objeto)`

```python
produtos_eletronicos.insert(2, "Telão")
print(produtos_eletronicos)
```

## 2. pop()

O `pop()` remove um elemento da lista.

Quando usamos apenas `pop()` sem passar nenhum argumento, ele remove o último elemento.

No nosso exemplo, o último produto será removido.

Também não podemos passar o nome do objeto diretamente:

`produtos_eletronicos.pop("Geladeira")`

Isso daria erro porque o `pop()` trabalha com índice, e não com o valor do objeto.

```python
produtos_eletronicos.pop()
print(produtos_eletronicos)
```

## 3. index()

O `index()` retorna o índice onde um determinado valor foi encontrado.

Quando existem valores repetidos, ele retorna a primeira ocorrência encontrada.

Também podemos informar um intervalo de busca.

A estrutura pode ser:

`lista.index(objeto)`

ou:

`lista.index(objeto, inicio, fim)`

```python
posicao_computador = produtos_eletronicos.index("Computador")
print(posicao_computador)
```

Agora vamos criar uma lista numérica para observar melhor o comportamento do `index()` quando existem valores repetidos.

```python
lista_num = [0,5,3,6,8,3,1,4,6,1,4,3,5,7,2]
```

Primeiro verificamos o tamanho da lista.

Depois procuramos o número `1` utilizando:

`lista_num.index(1, 2, 15)`

Nesse caso:

- `1` é o valor procurado;
- `2` é o índice onde a busca começa;
- `15` é o limite da busca.

O número `1` aparece mais de uma vez, mas o `index()` retorna a primeira ocorrência encontrada dentro desse intervalo.

```python
print(len(lista_num))
posicao_num = lista_num.index(1, 2, 15)
print(posicao_num)
```

## 4. count()

O `count()` serve para contar quantas vezes determinado elemento aparece dentro de uma lista.

Aqui queremos descobrir quantas vezes o número `1` aparece.

```python
print(lista_num.count(1))
```

## 5. sort()

O `sort()` ordena os elementos da própria lista.

Por padrão, ele organiza os valores em ordem crescente.

Um detalhe importante é que o `sort()` modifica a lista original.

```python
lista_num.sort()
print(lista_num)
```

## 6. reverse()

O `reverse()` inverte a ordem atual dos elementos da lista.

Ele não organiza em ordem crescente ou decrescente. Apenas pega a ordem existente e a inverte.

Vamos usar outra lista para visualizar isso melhor.

```python
animais = ["Cachorro" , "Papagaio" , "Gato" , "Elefante" , "Jacaré"]
print(f'Antes do Reverse: {animais}')
animais.reverse()
print(f'Depois do Reverse: {animais}')
```

## 7. extend()

O `extend()` adiciona os elementos de outro iterável ao final da lista original.

Aqui vamos usar duas listas de estados.

Diferente do `append()`, o `extend()` não coloca a segunda lista inteira como um único elemento. Ele adiciona os elementos dela na primeira lista.

```python
uf1 = ["RN" , "CE" , "PB" , "RJ"]
uf2 = ["SP" , "BA" , "ES" , "AM"]

uf1.extend(uf2)
print(uf1)
```
