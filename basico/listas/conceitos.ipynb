# Listas em Python

As listas são estruturas de dados utilizadas para armazenar vários elementos dentro de uma única variável.

Em Python, uma lista é criada utilizando colchetes `[]`, e seus elementos são separados por vírgulas.

Uma característica importante das listas é que elas são **mutáveis**, ou seja, podemos adicionar, remover ou modificar seus elementos depois que a lista foi criada.

Neste notebook utilizaremos como exemplo uma lista de produtos eletrônicos de uma loja.

```python
produtos_eletronicos = [
    "Computador",
    "Celular",
    "Televisão",
    "PlayStation",
    "Geladeira",
    "Nobreak"
]
```

## len()

O `len()` retorna a quantidade de elementos da lista.

```python
print(len(produtos_eletronicos))
```

## Índices

Os índices começam em 0.

Então:
- Computador = 0
- Celular = 1
- Televisão = 2
- PlayStation = 3

```python
print(produtos_eletronicos[0])
print(produtos_eletronicos[3])
```

## Slicing

Podemos pegar apenas uma parte da lista.

No exemplo abaixo começa no índice 0 e vai até antes do 3.

```python
print(produtos_eletronicos[0:3])
```

## Índices negativos

Também podemos acessar a lista de trás para frente.

O `-1` sempre representa o último elemento.

```python
print(produtos_eletronicos[-1])
```

## Step

No slicing também podemos definir de quantos em quantos elementos queremos avançar.
lista[inicio : fim : step]

```python
print(produtos_eletronicos[0:5:2])
```

## Alterando um elemento

Listas são mutáveis, então podemos alterar um valor depois que a lista já foi criada.

```python
produtos_eletronicos[1] = "Smartphone"

print(produtos_eletronicos)
```

## append()

O `append()` adiciona um novo elemento no final da lista.

O valor que colocamos dentro do `append()` é o argumento passado para o método.

```python
produtos_eletronicos.append("Notebook")

print(produtos_eletronicos)
```

## remove()

O `remove()` remove um elemento da lista pelo valor.

```python
produtos_eletronicos.remove("Geladeira")

print(produtos_eletronicos)
```

## Percorrendo a lista

Podemos usar o `for` para passar por cada elemento da lista.

A variável `produto` recebe um elemento por vez.

```python
for produto in produtos_eletronicos:
    print(produto)
```
