## Gabarito da questão 4

Primeiro calculamos os totais. Depois, `zip()` permite percorrer os nomes e os totais correspondentes ao mesmo tempo.

````python
lojas = ['Loja A', 'Loja B', 'Loja C']
vendas = [
    [1200, 1500, 1100, 1200],
    [900, 1100, 1300, 1000],
    [2000, 1800, 2100, 1900]
]

totais = []
for vendas_da_loja in vendas:
    totais.append(sum(vendas_da_loja))

for loja, total in zip(lojas, totais):
    print(f'{loja}: R$ {total:.2f}')
````

## Gabarito da questão 5

`enumerate()` fornece ao mesmo tempo o índice e o valor que está sendo percorrido.

````python
temperaturas = [
    [23, 25, 27, 26],
    [22, 28, 29, 24],
    [25, 31, 30, 27]
]

maior = temperaturas[0][0]
linha_maior = 0
coluna_maior = 0

for indice_linha, linha in enumerate(temperaturas):
    for indice_coluna, temperatura in enumerate(linha):
        if temperatura > maior:
            maior = temperatura
            linha_maior = indice_linha
            coluna_maior = indice_coluna

print(f'Maior temperatura: {maior}°C — linha {linha_maior}, coluna {coluna_maior}')
````

## Gabarito da questão 6

O fatiamento `aluno[1:]` cria uma lista contendo todos os elementos a partir do índice 1, ou seja, somente as notas.

````python
turma = [
    ['Ana', 8.0, 7.5, 8.5],
    ['Bruno', 6.0, 5.5, 7.0],
    ['Carla', 9.0, 8.5, 10.0],
    ['Diego', 7.0, 7.0, 6.5]
]

for aluno in turma:
    nome = aluno[0]
    notas = aluno[1:]
    media = sum(notas) / len(notas)
    situacao = 'Aprovado' if media >= 7 else 'Reprovado'
    print(f'{nome} — média: {media:.1f} — {situacao}')
````

## Gabarito da questão 7

Para cada índice de coluna da matriz original, criamos uma nova linha e adicionamos a ela os itens daquela coluna.

````python
original = [
    [1, 2, 3],
    [4, 5, 6]
]

transposta = []
for coluna in range(len(original[0])):
    nova_linha = []
    for linha in range(len(original)):
        nova_linha.append(original[linha][coluna])
    transposta.append(nova_linha)

print(transposta)
````

## Gabarito da questão 8

Contamos os estados de cada assento e guardamos o total de lugares livres de cada fileira. O método `index()` retorna a primeira posição do maior total, resolvendo também a regra de desempate.

````python
assentos = [
    ['L', 'X', 'L', 'X', 'L'],
    ['X', 'X', 'L', 'L', 'X'],
    ['L', 'L', 'L', 'X', 'L'],
    ['X', 'L', 'X', 'L', 'X']
]

total_livres = 0
total_ocupados = 0
livres_por_fileira = []

for fileira in assentos:
    livres_na_fileira = 0
    for assento in fileira:
        if assento == 'L':
            total_livres += 1
            livres_na_fileira += 1
        else:
            total_ocupados += 1
    livres_por_fileira.append(livres_na_fileira)

maior_quantidade = max(livres_por_fileira)
melhor_fileira = livres_por_fileira.index(maior_quantidade) + 1

print(f'Assentos livres: {total_livres}')
print(f'Assentos ocupados: {total_ocupados}')
print(f'Livres por fileira: {livres_por_fileira}')
print(f'Fileira com mais lugares livres: {melhor_fileira}')
````
