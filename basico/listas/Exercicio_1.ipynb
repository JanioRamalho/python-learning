## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]
```

```python
### Vamos começar analisando o que foi pedido de forma bem direta
'''
1. Maior valor de vendas
2. Pior valor de vendas
'''
# Poderiamos fazer apenas observando as listas e dizer qual a resposta que a questão pede.
# Porém, se fosse uma lista de milhares de elementos?? Seria impossível analisar no olho


meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = ([25000, 29000, 22200, 17750, 15870, 19900])
vendas_2sem = ([19850, 20120, 17540, 15555, 49051, 9650])

vendas_1sem.extend(vendas_2sem)
print(vendas_1sem)

print(f'Maior Venda: R$ {max(vendas_1sem):.2f}')
print(f'Menor Venda: R$ {min(vendas_1sem):.2f}')
```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
# Encontrando o melhor mês
maior_venda = max(vendas_1sem)
posicao_maior_venda = vendas_1sem.index(maior_venda)
melhor_mes = meses[posicao_maior_venda]

# Encontrando o pior mês
menor_venda = min(vendas_1sem)
posicao_menor_venda = vendas_1sem.index(menor_venda)
pior_mes = meses[posicao_menor_venda]

# Faturamento total do Ano 
faturamento = sum(vendas_1sem)


# % Representativa do melhor mês
# Representar em forma de Regra de 3
x = (maior_venda * 100) / faturamento

# Exibindo os resultados
print(f'O melhor mês do ano foi {melhor_mes} com {maior_venda} vendas.')
print(f'O pior mês do ano foi {pior_mes} com {menor_venda} vendas.')
print(f'O faturamento total foi de: {faturamento}')
print(f'{x:.2f} %')
```

## 3. Crie uma lista com o top 3 valores de vendas do ano (sem fazer "no olho")

Dica: o método remove retira um item da lista.

```python
'''
Analisando a lista, é possível ver que ela está na ordem crescente. Se invertemos, pegamos os 3 maiores números em sequência, e 
deixamos a lista correr somente até o terceiro elemento
'''

top3 = sorted(vendas_1sem , reverse=True)
top3 = top3[:3]
print(top3)
```
