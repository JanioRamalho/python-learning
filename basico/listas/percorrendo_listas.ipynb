Exercício 1 — Controle de vendas

Você possui a seguinte lista:

`vendas = [850, 1200, 430, 1800, 950, 2100, 670]`

Faça um programa que percorra essa lista usando for e:

1. Mostre cada venda individualmente.
2. Conte quantas vendas foram de **R$ 1.000 ou mais**.
3. Calcule o valor total de todas as vendas.
4. Ao final, exiba o total vendido e a quantidade de vendas acima ou iguais a **R$ 1.000**.

```python
vendas = [850 , 1200, 430, 1800, 950 , 2100, 670]

# Vou deixar a lista em ordem crescente, somente para fácil visualização, não altera o resultado final

vendas.sort()
print(vendas)

# Vamos percorrer cada item da lista
for indice , preco in enumerate(vendas):
    print(f"R$ {preco:.2f}")




# Vendas Maior ou igual a 1000
preco_alto = []

for preco in vendas:
    if preco >= 1000:
        preco_alto.append(preco)
print(f'{len(preco_alto)} Preços são maiores ou igual a 1000 --> {preco_alto}')




#Calcular Valor Total das Vendas
valor_total = 0
for preco in vendas:
    valor_total += preco
print(f'O Valor Total foi de: R$ {valor_total:.2f}')






```

## Exercício 2 — Análise de estoque

Você possui duas listas:

`produtos = ["Notebook", "Mouse", "Teclado", "Monitor", "Headset"]`


`estoque = [3, 15, 8, 2, 12]`

Cada posição de produtos corresponde à mesma posição em estoque.

Ou seja:

Notebook -> 3
Mouse -> 15
Teclado -> 8
Monitor -> 2
Headset -> 12

Faça um programa que percorra as listas e mostre:

1. o nome de cada produto;
2. sua quantidade em estoque;
3. se o estoque for menor que 5, mostre **"ESTOQUE BAIXO"**;
4. ao final, informe quantos produtos estão com estoque baixo.

Olhando o exemplo a primeira vista parece algo mais complicado do que de fato é;

- A idéia aqui não é colocar primeiro o Nome do produto em uma linha e a Quantidade do estoque em outra, e sim, unificar essas duas coisas em uma única linha. A saída ficaria mais ou menos assim:

1. Notebook: 3 unidades - ESTOQUE BAIXO
2. Mouse: 15 unidades
3. Teclado: 8 unidades
4. Monitor: 2 unidades - ESTOQUE BAIXO
5. Headset: 12 unidades

6. Quantidade de produtos com estoque baixo: 2

```python
produtos = ["Notebook", "Mouse", "Teclado", "Monitor", "Headset"]
estoque = [3, 15, 8, 2, 12]
estoque_baixo = 0

# Para Relacionarmos duas listas podemos usar o zip()
# zip() é uma função do Python que serve para percorrer duas ou mais listas ao mesmo tempo,relacionando os
# elementos que estão na mesma posição.

for produto, quantidade in zip(produtos, estoque):
    if quantidade <= 5: 
        estoque_baixo += 1
        print(f'{produto} -> {quantidade} Unidades  / Estoque Baixo')
        
    else:
        print(f'{produto} -> {quantidade} Unidades')

print(f'\nQuantidade de produtos com estoque baixo: {estoque_baixo}')
    

```

O código ficou bem mais curto e mais agradável visualmente do que se fizessemos um linha produtos e em outra quantidade
