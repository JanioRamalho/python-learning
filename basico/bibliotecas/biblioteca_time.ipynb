import time

```python
import time 
```

- Marco Zero: EPOCH(01 de Janeiro de 1970 às 00:00:00)
- time() retorna quantos segundos se passaram desde a EPOCH

```python
print(time.time())
```

Isso pode ser muito útil para medir o tempo que um codigo leva pra rodar

```python
data_inicial = time.time()
# Rodo o codigo que eu quero saber quanto tempo demora
for i in range(1000000):
    pass

# Pega o tempo final

data_final = time.time()

print(f'O tempo de execução do código foi de {data_final - data_inicial:.4f} segundos')
```

- ctime() retorna a data em string


```python
data_hoje = time.ctime()
print(f'A data de hoje é: {data_hoje}')
print(type(data_hoje))
```

Fazer o programa esperar alguns segundos (Muito útil quando temos que esperar um programa ou uma página carregar)

```python
# Abrir o navegador
print("Abrir navegador")

time.sleep(5) # Tempo em segundos

#Entrar no sistema
print("Entrar no sistema")
```

Pegar informações de dias, hora, minuto, segundo, tudo detalhado:

- gmtime()  
- localtime()
  

```python
data_global = time.gmtime()
data_local = time.localtime()

print(f'Data global: {data_global}')
print(f'Data local: {data_local}')
```

```python
dia_de_hoje = time.gmtime()
print(f"Hoje é dia {dia_de_hoje.tm_mday}/{dia_de_hoje.tm_mon}/{dia_de_hoje.tm_year}")

mes_hoje = time.gmtime()
print(f"Hoje é o mês {mes_hoje.tm_mon}")
```
