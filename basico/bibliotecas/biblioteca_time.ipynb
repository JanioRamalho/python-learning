- Marco Zero: EPOCH(01 de Janeiro de 1970 às 00:00:00)
- time() retorna quantos segundos se passaram desde a EPOCH

```python
print(time.time())
```

Isso pode ser muito útil para medir o tempo que um codigo leva pra rodar

```python
data_inicial = time.time()
# Rodo o codigo que eu quero saber quanto tempo demora
for i in range(1000000):
    pass

# Pega o tempo final

data_final = time.time()

print(f'O tempo de execução do código foi de {data_final - data_inicial:.4f} segundos')
```

- ctime() retorna a data em string


```python
data_hoje = time.ctime()
print(f'A data de hoje é: {data_hoje}')
print(type(data_hoje))
```

Pegar informações de dias, hora, minuto, segundo, tudo detalhado:

- gmtime()  
- localtime()
  

```python
data_global = time.gmtime()
data_local = time.localtime()

print(f'Data global: {data_global}')
print(f'Data local: {data_local}')
```

```python
dia_de_hoje = time.gmtime()
print(f"Hoje é dia {dia_de_hoje.tm_mday}/{dia_de_hoje.tm_mon}/{dia_de_hoje.tm_year}")

mes_hoje = time.gmtime()
print(f"Hoje é o mês {mes_hoje.tm_mon}")
```

Agora, para trabalhar com formatações de datas, consulte esse link: https://docs.python.org/3/library/time.html#time.strtime

E para formatar datas para serem exibidas em uma mensagem do jeito que a gente quer?


```python
import time 

data_loca2 = time.localtime()
data_texto = time.strftime("%d/%m/%Y %A", data_loca2)

print(f"Hoje é dia {data_texto}")
```

E se eu quiser o texto em Português (do Mês e dia da Semana). Ai precisamos dizer para o Python o seu "local", usando a biblioteca locale. Link: https://docs.python.org/pt-br/3/library/locale.html

```python
import locale

# locale.setlocale("Tipo de localização", "Qual a lingua/ local que você quer usar")

locale.setlocale(locale.LC_TIME, "pt_BR.UTF-8")
data_texto = time.strftime("%d/%m/%Y %A %B", data_loca2)

print(f"Hoje é dia {data_texto}")
```

E para ler arquivos com datas formatadas de maneiras diferentes?

```python
lista_datas = ["01 de Junho de 2026", "18 de Julho de 2026", "03 de Agosto de 2026", "15 de Setembro de 2026", "30 de Outubro de 2026"]

for data in lista_datas:
    data_estruturada = time.strptime(data, "%d de %B de %Y" )
    print(f"Data estruturada: {data_estruturada}")
```
